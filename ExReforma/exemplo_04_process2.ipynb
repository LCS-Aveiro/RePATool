Example 4: Large-Scale Process Mining on eCommerce Data

This example demonstrates how **ReForma** handles real-world, large-scale datasets. We use the [eCommerce behavior data from a multi-category store](https://www.kaggle.com/datasets/mkechinov/ecommerce-behavior-data-from-multi-category-store/data) dataset from Kaggle to mine, build, and train a Reconfigurable Formal Automaton.

```python
import time
import subprocess
import os
import re
import pandas as pd
from reforma import ReForma

print("Importing libraries...")
t_start_imports = time.perf_counter()

t_end_imports = time.perf_counter()
print(f"⏱️ Libraries loaded in {t_end_imports - t_start_imports:.2f} seconds.\n")

print("Loading e-commerce data (CSV)...")
t_start_csv = time.perf_counter()
df = pd.read_csv('2019-Oct.csv', usecols=['event_time', 'event_type', 'user_session'])
t_end_csv = time.perf_counter()
print(f"⏱️ CSV read in {t_end_csv - t_start_csv:.2f} seconds. ({len(df)} rows processed)\n")

model = ReForma()

print("Discovering and building the RePA automaton...")
t_start_traces = time.perf_counter()
traces = model.create_from_dataframe(
    df,
    session_col='user_session',
    event_col='event_type',
    time_col='event_time',
    model_name="ECommerce_Mined",
    save_model_path="my_ecommerce_model.rta",
    save_traces_path="my_training_sessions.txt"
)
t_end_traces = time.perf_counter()
print(f"⏱️ Automaton built and {len(traces)} traces generated in {t_end_traces - t_start_traces:.2f} seconds.\n")

print("Generating the graph...")
model.save_image_plt('ecommerce_graph.png') 
print("✔️ Graph saved to 'ecommerce_graph.png'\n")

print("Training the model with the extracted sessions...")
t_start_train = time.perf_counter()
model.train(traces)
t_end_train = time.perf_counter()
print(f"⏱️ Training completed in {t_end_train - t_start_train:.2f} seconds.\n")

t_total = t_end_train - t_start_imports
print(f"🚀 Script finished successfully! Total time: {t_total:.2f} seconds.")

with open("my_trained_model.rta", "w", encoding="utf-8") as f:
    f.write(model.source)
print("✔️ Final model saved to 'my_trained_model.rta'")
```
